```python
# Удаляем NaN для колонок name, genre
new_games = games.dropna(subset=['name', 'genre'])
```

```python
# Убираем все строки с Nan в дате
new_games = new_games.dropna(subset=['year_of_release'])
```

```python
# Замена пропусков на средние значения в колонках eu_sales и jp_sales
def mean_group_sales(row, column):
    if pd.isna(row[column]):
        group = new_games[(new_games['platform'] == row['platform']) & 
                             (new_games['year_of_release'] == row['year_of_release'])]
        return group[column].mean()
    else:
        return row[column]
    
columns = ['eu_sales', 'jp_sales']
for column in columns:
    new_games[column] = new_games.apply(lambda row: mean_group_sales(row, column), axis=1)
```

```python
# Замена значений NaN в 'critic_score' на -1
new_games['critic_score'] = new_games['critic_score'].fillna(-1)
```

```python
# Замена значений NaN в 'user_score' на -1
new_games['user_score'] = new_games['user_score'].fillna(-1)
```

Все рейтинги категорированы на `Нет оценки`, `Низкая оценка`, `Средняя оценка`, `Высокая оценка`.

- После категоризации данных проверьте результат: сгруппируйте данные по выделенным категориям и посчитайте количество игр в каждой категории.

```python
# Считаем общую сумму продаж
df_actual['total_sales'] = df_actual['na_sales'] + df_actual['eu_sales'] + df_actual['jp_sales'] + df_actual['other_sales']
```

```python
# Создаем функцию для группировки
def result_category(column_name):
    result = df_actual.groupby(column_name).agg({'name': 'count','total_sales': 'sum'})
    result = result.sort_values(by='name', ascending=False).reset_index()
    result['count_share'] = round(result['name']/ result['name'].sum() *100,1)
    result = result.rename(columns={'name': 'count'})
    result['total_sales'] = round(result['total_sales'], 1)
    result['avg_sales_per_game'] = round((result['total_sales'] / result['count']) * 1000)
    return result
```
